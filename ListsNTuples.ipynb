```python
#2. Accessing Negative Indexing

#Python allows negative indexing, which means counting from the end of the list.

nums = [10, 20, 30, 40]

"""
| Index (negative) | -4 | -3 | -2 | -1 |
| ---------------- | -- | -- | -- | -- |
| Value            | 10 | 20 | 30 | 40 |
"""

print(nums[-1])  # 40 (last)
print(nums[-2])  # 30 (second last)
print(nums[-4])  # 10 (first)

"""
⚡ Why useful?
You don’t have to know the exact list length — just use negative indices to get items from the end.
"""
```

```python
#dynamic slicing

#We’ll assume a general 2D list matrix:

matrix = [
  [...],  # row 0
  [...],  # row 1
  ...
]


#Get the number of rows and columns

num_rows = len(matrix)
num_cols = len(matrix[0])  # assume all rows have the same length

"""
2️⃣ Compute the middle indices

Suppose we want a middle 4×4 block (or any size block_rows × block_cols):
"""

block_rows = 4
block_cols = 4

# Start indices (middle)
start_row = (num_rows - block_rows) // 2
start_col = (num_cols - block_cols) // 2


# End indices
end_row = start_row + block_rows
end_col = start_col + block_cols

"""
This centers the block automatically, even if num_rows or num_cols is unknown.

Works for both even and odd dimensions.
"""

"""
Slice the middle block dynamically
"""

middle_block = [row[start_col:end_col] for row in matrix[start_row:end_row]]
```

```python
#dynamic slicing with 10x10 matrix

matrix_10x10 = [list(range(i*10+1, i*10+11)) for i in range(10)]

num_rows = len(matrix_10x10)   # 10
num_cols = len(matrix_10x10[0])  # 10

block_rows = 4
block_cols = 4

start_row = (num_rows - block_rows) // 2  # 3
start_col = (num_cols - block_cols) // 2  # 3
end_row = start_row + block_rows          # 7
end_col = start_col + block_cols          # 7

middle_block = [row[start_col:end_col] for row in matrix_10x10[start_row:end_row]]

middle_block
```

```python
#Sort based on perticular key column

data = [
    ["apple", 4, 128, 900],
    ["samsung", 12, 128, 800],
    ["lg", 8, 64, 700]
]

# Define a function that returns the column you want to sort by
def get_2col(row):
    return row[1]  # RAM column

# Sort using the function
data.sort(key=get_2col)
print(data)
```

```python
# Sort using the function in descending order
data.sort(key=get_2col, reverse=True)
print(data)
```

| **Method**                      | **Description**                                                                     | **Example**                                               |
| ------------------------------- | ----------------------------------------------------------------------------------- | --------------------------------------------------------- |
| `append(x)`                     | Adds a single element `x` to the end of the list                                    | `fruits.append("mango")` → `['apple', 'banana', 'mango']` |
| `extend(iterable)`              | Adds all elements from another list (or iterable) to the end                        | `fruits.extend(["grape", "kiwi"])`                        |
| `insert(i, x)`                  | Inserts `x` at index `i` (shifts right)                                             | `fruits.insert(1, "orange")`                              |
| `remove(x)`                     | Removes the **first occurrence** of element `x`                                     | `fruits.remove("apple")`                                  |
| `pop([i])`                      | Removes and returns the element at index `i`. If no index, removes the last element | `fruits.pop(2)`                                           |
| `clear()`                       | Removes all elements (makes list empty)                                             | `fruits.clear()`                                          |
| `index(x[, start[, end]])`      | Returns the **index** of the first occurrence of `x`                                | `fruits.index("banana")`                                  |
| `count(x)`                      | Returns how many times `x` appears in the list                                      | `nums.count(10)`                                          |
| `sort(key=None, reverse=False)` | Sorts the list **in place**                                                         | `nums.sort()` or `nums.sort(reverse=True)`                |
| `reverse()`                     | Reverses the list **in place**                                                      | `fruits.reverse()`                                        |
| `copy()`                        | Returns a **shallow copy** of the list                                              | `new_list = fruits.copy()`                                |
