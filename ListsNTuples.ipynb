```python
# Store the data in a variable
mobile_store = [
    {"brand": "apple", "ram(GB)": 4, "memory(GB)": 128, "price($)": 900},
    {"brand": "samsung", "ram(GB)": 12, "memory(GB)": 128, "price($)": 899},
    {"brand": "lg", "ram(GB)": 8, "memory(GB)": 64, "price($)": 600},
    {"brand": "apple", "ram(GB)": 8, "memory(GB)": 128, "price($)": 1000},
]

# Print the data
for row in mobile_store:
    print(row)
```

```python
#LIST
"""
Let's create a variable that will store all the brand names given in the above table. this can be acheived using
the "list" data structure in Python.
Lists are used to store the data items inside square brackets [] where each data item is seperated by a comma(,)

"""
#Create a list
brand_list=["Apple","Samsung","LG","Apple"]
brand_list
```

```python
#Check the data type of the variable 'brand_list'
type(brand_list)
```

```python
#Q Print the third item in the list ram_list
"""
Indexing in List is done by specifying the index inside square brackets.in python,indexing starts 
with 0. That is, the first item has the index 0, the second has the index 1 and so on

"""
#Indexing to get the third item in ram_list

print(ram_list[2])

```

```python
#Print the first three items from the list price_list
"""
We can also select a range of indices to access more than 1 item. This can be done by using a colon (:)
between the start and stop index 
"""
#First three items from the price list
print(price_list[0:3])

```
