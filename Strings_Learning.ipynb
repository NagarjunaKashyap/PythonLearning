Strings

```python
"""
Python String Data Type
"""
s1='Hello'
s2='World'
print(s1,s2)
```

```python
#Creating Strings

str1 = 'Python'
str2 = "Python"
print(str1,str2)
```

```python
#Triple quotes for multiline strings:
str3 = """This is
a multiline
string"""
print(str3)
```

```python
#Using str() to convert other types:
num = 100
str_num = str(num)
print(str_num, type(str_num))  # '100' <class 'str'>
```

```python
#Accessing Characters (Indexing)
text = "Python"
print(text[0])   # P (first character)
print(text[5])   # n (last character)
print(text[-1])  # n (negative indexing, last character)
```

```python
#String Slicing
#You can extract parts of a string using start:stop:step:

text = "Python"

print(text[0:4])   # 'Pyth' (0 to 3)
print(text[:4])    # 'Pyth' (start defaults to 0)
print(text[2:])    # 'thon' (till end)
print(text[::2])   # 'Pto' (every 2nd character)
print(text[::-1])  # 'nohtyP' (reverse string)
```

```python
#String Concatenation and Repetition
s1 = "Hello"
s2 = "World"
print(s1 + " " + s2)  # Hello World

print(s1 * 3)  # HelloHelloHello
```

```python
#Common String Methods
#Python strings have built-in methods. Let’s categorize them.
#a) Case Conversion
s = "python"
print(s.upper())   # 'PYTHON'
print(s.lower())   # 'python'
print(s.title())   # 'Python'
print(s.capitalize())  # 'Python'
print(s.swapcase())    # 'PYTHON' -> 'python'
```

```python
#b) Searching & Checking
s = "Hello World"

print(s.startswith("Hello"))  # True
print(s.endswith("World"))    # True
print(s.find("o"))            # 4 (first occurrence)
print(s.count("l"))           # 3 (count of 'l')
print("123".isdigit())        # True
print("abc".isalpha())        # True
print("abc123".isalnum())     # True
print(" ".isspace())           # True
```

```python
#c) Trimming & Replacing

s = "   Python   "
print(s.strip())       # 'Python' (remove spaces)
print(s.lstrip())      # 'Python   ' (left only)
print(s.rstrip())      # '   Python' (right only)

str = "I like Java"
print(str.replace("Java", "Python"))  # 'I like Python'
```

```python
#d) Splitting & Joining
s = "Python is awesome"

# Split into list
words = s.split()           # ['Python', 'is', 'awesome']
print(words)

# Split by character
chars = s.split("o")        # ['Pyth', 'n is awes', 'me']
print(chars)

# Join list into string
lst = ["Python", "is", "fun"]
joined = "-".join(lst)      # 'Python-is-fun'
print(joined)
```

```python
#e) Formatting Strings
name = "Alice"
age = 25
print(f"My name is {name} and I am {age} years old")
```

```python
#Using format():
print("My name is {} and I am {} years old".format(name.upper(), age))
print("My name is {0} and I am {1} years old".format(name.title(), age))
print("My age is {age} and name is {name}".format(name=name, age=age))
```

```python
#Using % formatting (old style):
print("My name is %s and I am %d years old" % (name, age))
```

```python
#f) Escape Characters

s = "Hello\nWorld"  # \n -> new line
print(s)

s = "She said \"Hi\""
print(s)

s = "C:\\Users\\Alice"  # Backslash
print(s)
```

Common escapes:

\n → newline

\t → tab

\\ → backslash

\' → single quote

\" → double quote

```python
#g) Checking String Properties
s = "Python3"
print(s.isalnum())  # True
print(s.isalpha())  # False
print(s.isdigit())  # False
print(s.islower())  # False
print(s.isupper())  # False
print(s.istitle())  #True
```

6️⃣Advanced Examples

1.Reverse a string:

```python
s = "Python"
rev = s[::-1]
print(rev)  # nohtyP
```

Count vowels in string:

```python
s = "Hello World"
vowels = "aeiouAEIOU"
count = 0
for char in s:
    if char in vowels:
        count += 1
print(count)  # 3
```

```python
#Check palindrome:

s = "madam"
if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```
