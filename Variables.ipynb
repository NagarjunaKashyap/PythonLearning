```python
"""
Type Conversion: The process of converting one data type to another data type is called type conversion.

in Python, we can perform two types of conversion.

1.Implicit conversion - Here, python automatically converts data types from one to another
2.explicit comversion - here, The user can convert the data type of a variable to the required 
data type by using the built - in functions
int(),float(),str(),etc.
"""

#Store the price (in dollars) of the mobile in variable 'price'
price=900
#calculate the discount amount
discount=price*4.5/100
#calculate the discounted price
discount_price=price-discount

#print the discounted price
print("The discounted price of the iPhone is " + str(discount_price),sep=" ")
```

```python
#Now check the data type of the variable 'discount' and 'price'
print("The data type of the varible price is",type(price))
print("The data type of the varible discount is",type(discount))

"""
As we can see, the data type of 'discount' is float as python is impiclitly converting the lower data type to 
higher data type to avoid data loss
"""
```
