4. Tuple Immutability

```python
#Tuples cannot be changed:

t = (10, 20, 30)
t[0] = 100  ❌ TypeError
```

```python
#However, if a tuple contains mutable objects (like a list), those can be modified:
t = (1, [2, 3], 4)
t[1][0] = 200
print(t)  # (1, [200, 3], 4)
```

5. Tuple Iteration and Membership

```python
#Iteration

t = ('apple', 'banana', 'cherry')
for item in t:
    print(item)

#Membership
print('banana' in t) #True
print('mango' in t)  #False
print('orange' not in t) #True
    
```

6. Tuple Methods

```python
#Tuples have only two built-in methods:

t=(1,2,2,3,2,4)
print(t.count(2)) # Times repeated
```

```python
#index(value, start=0, end=len(tuple)) → returns index of first match
t = (10, 20, 30, 20, 40)
print(t.index(20))     #1
print(t.index(10))     #0
print(t.index(20, 2))  # 3
```

#7. Tuple Unpacking

```python

#You can assign tuple elements to multiple variables at once.
t=(100,200,300)
a,b,c=t
print(a)
print(b)
print(c)
print(t[1])
print(t[0:2])
```

```python
t = (10, 20, 30, 40, 50)
a, *b, c = t
print(a)  # 10
print(b)  # [20, 30, 40]
print(c)  # 50
```

| Variable | What it gets   | Why                                                                                              |
| -------- | -------------- | ------------------------------------------------------------------------------------------------ |
| `a`      | `10`           | It takes the **first** element of the tuple.                                                     |
| `c`      | `50`           | It takes the **last** element of the tuple.                                                      |
| `*b`     | `[20, 30, 40]` | The `*` operator “soaks up” **all remaining middle values**. They are collected into a **list**. |


💡 Key point

The * (star) before a variable name in unpacking means:

“Collect the rest of the unpacked values into a list.”

This works not only with tuples but also with lists and any iterable.

🧠 More examples
Example 1: Basic list unpacking

```python
numbers = [1, 2, 3, 4, 5, 6]
x, *y = numbers
print(x)  # 1
print(y)  # [2, 3, 4, 5, 6]
```

Example 2: Star in the middle

```python
x, *y, z = [1, 2, 3, 4, 5]
# x = 1, y = [2, 3, 4], z = 5
print(x)
print(y)
print(z)
```

#Example 3: Star at the start

```python
*a, b = [10, 20, 30]
# a = [10, 20], b = 30
print(a)
```

```python
#Example 4: Nested unpacking
t = (1, (2, 3, 4), 5)
a, (*b), c = t
# a = 1, b = [2, 3, 4], c = 5
```

```python
#Correct way of unpacking
#✅ Option 1 — Two-step unpacking

t = (1, (2, 3, 4), 5)
a, inner, c = t
b = [*inner]   # convert inner tuple to a list if you want
print(a)  # 1
print(b)  # [2, 3, 4]
print(c)  # 5
```

```python
#✅ Option 2 — Use slicing on the inner element

#If you specifically want to use * logic dynamically:

#t = (1, (2, 3, 4), 5)
a, (b1, b2, b3), c = t
print(a)   # 1
print(b1)  # 2
print(b2)  # 3
print(b3)  # 4
print(c)   # 5
```

🧰 11. Real-life Scenarios
✅ Returning multiple values from a function:

```python
def get_stats(numbers):
    return (min(numbers), max(numbers), sum(numbers)/len(numbers))

result = get_stats([10, 20, 30, 40])
print(result)  # (10, 40, 25.0)
```

Swapping values without a temporary variable:

```python
a, b = 5, 10
a, b = b, a
print(a, b)  # 10 5
```

✅ Storing records:

student = ("John", 21, "Physics")
name, age, subject = student
print(name, age, subject)


✅ Iterating over list of tuples:

```python
students = [("John", 85), ("Alice", 90), ("Bob", 75)]
for name, score in students:
    print(f"{name} scored {score}")
```

```python
#Sorting

data = [(2, 'b'), (1, 'a'), (3, 'c')]
sorted_data = sorted(data, key=lambda x: x[1])
print(sorted_data)  # [(1, 'a'), (2, 'b'), (3, 'c')]
```

 Tuple Conversion and Casting

```python
#convert list to tuple

t = tuple([1, 2, 3])
print(t)
```

```python
#Convert tuple → list (to modify it):
l = list((1, 2, 3))
l.append(4)
t = tuple(l)
print(t)  # (1, 2, 3, 4)
```

```python
names=('Alice','Bob','Charlie')
scores=(85,90,95)

zipped=tuple(zip(names,scores))

print(zipped)
```

```python
#You can also unzip
n,s=zip(*zipped)
print(n)
print(s)
```

```python
#12. Sorting and Custom Keys
data = [(1, 'c'), (1, 'a'), (1, 'b')]
print(sorted(data))  # [(1, 'a'), (1, 'b'), (1, 'c')]

```

```python
#You can specify which part to sort by (like earlier):
sorted(data, key=lambda x: x[1])
```

```python
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(index, fruit)
# (0, 'apple'), (1, 'banana'), (2, 'cherry')
```

```python
#14. Tuple Slicing Tricks
t = (0, 1, 2, 3, 4, 5)
print(t[::2])  # (0, 2, 4)
print(t[::-1]) # reversed tuple
```

```python
#15. Tuples in Sets

#Tuples can be elements of a set (since they’re immutable):

points = {(0, 0), (1, 2), (2, 4)}
print(points)
```
